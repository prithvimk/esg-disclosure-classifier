```python
import pandas as pd
```

```python
file = pd.read_excel("../disclosure_standards/paragraph_mapping.xlsx", sheet_name=None)
```

```python
dfs = {}
excel_file_path = 'esrs_paragraph_mapping.xlsx'
with pd.ExcelWriter(excel_file_path, engine='openpyxl') as writer:
    for sheet_name, df in file.items():
        df_name = sheet_name.replace(' ','').lower()
        df.dropna(how='all', inplace=True)
        df.paragraph = df.paragraph.fillna(method='ffill')
        df.details = df.details.fillna(method='ffill')
        df.to_excel(writer, sheet_name=df_name, index=False)
        dfs[df_name] = df
```
